```python
from pathlib import Path

BASE_DIR = Path("/content/drive/MyDrive/ KrishiTrust_AI")
DATA_DIR = BASE_DIR / "datasets"

TRAIN_FILE = DATA_DIR / "fertilizer_train.csv"
TEST_FILE = DATA_DIR / "fertilizer_test.csv"

PREPROCESSOR_FILE = DATA_DIR / "fertilizer_preprocessor.joblib"
LABEL_ENCODER_FILE = DATA_DIR / "fertilizer_label_encoder.joblib"

print("Train:", TRAIN_FILE.exists())
print("Test:", TEST_FILE.exists())
print("Preprocessor:", PREPROCESSOR_FILE.exists())
print("Label encoder:", LABEL_ENCODER_FILE.exists())
```

```python
import pandas as pd

train_df = pd.read_csv(TRAIN_FILE)
test_df = pd.read_csv(TEST_FILE)

print("Training shape:", train_df.shape)
print("Testing shape:", test_df.shape)

print("\nTraining columns:")
print(train_df.columns.tolist())

print("\nTesting columns:")
print(test_df.columns.tolist())
```

```python
# ============================================================
# STEP 5.5 — DATA STRUCTURE VALIDATION
# ============================================================

TARGET = "Fertilizer"

NUMERICAL_FEATURES = [
    "Temperature",
    "Moisture",
    "Rainfall",
    "PH",
    "Nitrogen",
    "Phosphorous",
    "Potassium",
    "Carbon",
]

CATEGORICAL_FEATURES = [
    "Soil",
    "Crop",
]

EXPECTED_FEATURES = (
    NUMERICAL_FEATURES
    + CATEGORICAL_FEATURES
)

print("=" * 70)
print("STEP 5.5 — DATA STRUCTURE VALIDATION")
print("=" * 70)

# ------------------------------------------------------------
# Check columns
# ------------------------------------------------------------

print("\nTraining columns:")
print(train_df.columns.tolist())

print("\nTesting columns:")
print(test_df.columns.tolist())

# ------------------------------------------------------------
# Check required features
# ------------------------------------------------------------

missing_train = [
    col for col in EXPECTED_FEATURES + [TARGET]
    if col not in train_df.columns
]

missing_test = [
    col for col in EXPECTED_FEATURES + [TARGET]
    if col not in test_df.columns
]

print("\nMissing required training columns:", missing_train)
print("Missing required testing columns:", missing_test)

assert not missing_train
assert not missing_test

# ------------------------------------------------------------
# Check target
# ------------------------------------------------------------

print("\nTarget column:", TARGET)

print("\nTraining target distribution:")
print(train_df[TARGET].value_counts().sort_index())

print("\nTesting target distribution:")
print(test_df[TARGET].value_counts().sort_index())

# ------------------------------------------------------------
# Check missing values
# ------------------------------------------------------------

print("\nMissing values:")
print("Training:", train_df.isna().sum().sum())
print("Testing: ", test_df.isna().sum().sum())

assert train_df.isna().sum().sum() == 0
assert test_df.isna().sum().sum() == 0

# ------------------------------------------------------------
# Check Remark exclusion
# ------------------------------------------------------------

ML_FEATURES = EXPECTED_FEATURES

assert "Remark" not in ML_FEATURES

print("\nRemark included as feature:", "Remark" in ML_FEATURES)

# ------------------------------------------------------------
# Final
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STRUCTURE VALIDATION: PASS")
print("=" * 70)

print("\nFeatures used:")
for feature in ML_FEATURES:
    print(" -", feature)

print("\nTarget:")
print(" -", TARGET)

print("\nExcluded:")
print(" - Remark")
```

```python
# ============================================================
# STEP 5.6 — LEAKAGE-SAFE PREPROCESSING PIPELINE
# ============================================================

import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder

print("=" * 70)
print("STEP 5.6 — LEAKAGE-SAFE PREPROCESSING PIPELINE")
print("=" * 70)

# ------------------------------------------------------------
# Feature definitions
# ------------------------------------------------------------

NUMERICAL_FEATURES = [
    "Temperature",
    "Moisture",
    "Rainfall",
    "PH",
    "Nitrogen",
    "Phosphorous",
    "Potassium",
    "Carbon",
]

CATEGORICAL_FEATURES = [
    "Soil",
    "Crop",
]

TARGET = "Fertilizer"

# ------------------------------------------------------------
# Separate X and y
# ------------------------------------------------------------

X = train_df[
    NUMERICAL_FEATURES + CATEGORICAL_FEATURES
].copy()

y = train_df[TARGET].copy()

print("\nTraining feature shape:", X.shape)
print("Training target shape:", y.shape)

# ------------------------------------------------------------
# Create preprocessing transformer
# ------------------------------------------------------------

preprocessor_cv = ColumnTransformer(
    transformers=[
        (
            "numerical",
            StandardScaler(),
            NUMERICAL_FEATURES,
        ),
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False,
            ),
            CATEGORICAL_FEATURES,
        ),
    ],
    remainder="drop",
)

print("\nNumerical preprocessing:")
print("  StandardScaler")

print("\nCategorical preprocessing:")
print("  OneHotEncoder")
print("  handle_unknown='ignore'")

# ------------------------------------------------------------
# Test pipeline construction
# ------------------------------------------------------------

test_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor_cv),
    ]
)

print("\nPipeline created successfully.")

print("\nPipeline steps:")
print(test_pipeline.named_steps)

print("\n" + "=" * 70)
print("LEAKAGE-SAFE PIPELINE SETUP: PASS")
print("=" * 70)
```

```python
# ============================================================
# STEP 5.7 — 5-FOLD STRATIFIED CROSS-VALIDATION
# ============================================================

from sklearn.model_selection import StratifiedKFold

print("=" * 70)
print("STEP 5.7 — 5-FOLD STRATIFIED CROSS-VALIDATION")
print("=" * 70)

N_SPLITS = 5
RANDOM_STATE = 42

cv = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=RANDOM_STATE,
)

print("\nConfiguration:")
print(f"  Number of folds: {N_SPLITS}")
print(f"  Shuffle: True")
print(f"  Random state: {RANDOM_STATE}")
print(f"  Stratification target: {TARGET}")

print("\nFold sizes:")

for fold_number, (train_idx, val_idx) in enumerate(
    cv.split(X, y),
    start=1
):
    X_fold_train = X.iloc[train_idx]
    X_fold_val = X.iloc[val_idx]

    y_fold_train = y.iloc[train_idx]
    y_fold_val = y.iloc[val_idx]

    print(
        f"  Fold {fold_number}: "
        f"train={len(train_idx)}, "
        f"validation={len(val_idx)}"
    )

    # Verify every fertilizer class exists in both partitions
    train_classes = set(y_fold_train.unique())
    val_classes = set(y_fold_val.unique())
    all_classes = set(y.unique())

    assert train_classes == all_classes
    assert val_classes == all_classes

print("\nAll folds contain all fertilizer classes.")

print("\n" + "=" * 70)
print("STRATIFIED CROSS-VALIDATION SETUP: PASS")
print("=" * 70)
```

```python
# ============================================================
# STEP 5.8 — LOGISTIC REGRESSION BASELINE
# ============================================================

from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_validate
from sklearn.pipeline import Pipeline

print("=" * 70)
print("STEP 5.8 — LOGISTIC REGRESSION BASELINE")
print("=" * 70)

# ------------------------------------------------------------
# Model
# ------------------------------------------------------------

logistic_model = LogisticRegression(
    max_iter=2000,
    random_state=42,
)

# ------------------------------------------------------------
# Complete leakage-safe pipeline
# ------------------------------------------------------------

logistic_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            ColumnTransformer(
                transformers=[
                    (
                        "numerical",
                        StandardScaler(),
                        NUMERICAL_FEATURES,
                    ),
                    (
                        "categorical",
                        OneHotEncoder(
                            handle_unknown="ignore",
                            sparse_output=False,
                        ),
                        CATEGORICAL_FEATURES,
                    ),
                ],
                remainder="drop",
            ),
        ),
        (
            "model",
            logistic_model,
        ),
    ]
)

print("\nModel:")
print("  Logistic Regression")

print("\nConfiguration:")
print("  max_iter = 2000")
print("  random_state = 42")

# ------------------------------------------------------------
# Cross-validation scoring
# ------------------------------------------------------------

scoring = {
    "accuracy": "accuracy",
    "precision_macro": "precision_macro",
    "recall_macro": "recall_macro",
    "f1_macro": "f1_macro",
    "f1_weighted": "f1_weighted",
}

print("\nRunning 5-fold stratified cross-validation...")

cv_results_logistic = cross_validate(
    logistic_pipeline,
    X,
    y,
    cv=cv,
    scoring=scoring,
    n_jobs=-1,
    return_train_score=True,
)

# ------------------------------------------------------------
# Calculate summary
# ------------------------------------------------------------

logistic_summary = {
    "model": "Logistic Regression",

    "accuracy_mean":
        cv_results_logistic[
            "test_accuracy"
        ].mean(),

    "accuracy_std":
        cv_results_logistic[
            "test_accuracy"
        ].std(),

    "precision_macro_mean":
        cv_results_logistic[
            "test_precision_macro"
        ].mean(),

    "precision_macro_std":
        cv_results_logistic[
            "test_precision_macro"
        ].std(),

    "recall_macro_mean":
        cv_results_logistic[
            "test_recall_macro"
        ].mean(),

    "recall_macro_std":
        cv_results_logistic[
            "test_recall_macro"
        ].std(),

    "f1_macro_mean":
        cv_results_logistic[
            "test_f1_macro"
        ].mean(),

    "f1_macro_std":
        cv_results_logistic[
            "test_f1_macro"
        ].std(),

    "f1_weighted_mean":
        cv_results_logistic[
            "test_f1_weighted"
        ].mean(),

    "f1_weighted_std":
        cv_results_logistic[
            "test_f1_weighted"
        ].std(),
}

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("LOGISTIC REGRESSION — CROSS-VALIDATION RESULTS")
print("=" * 70)

print(
    f"\nAccuracy:       "
    f"{logistic_summary['accuracy_mean']:.4f} "
    f"± {logistic_summary['accuracy_std']:.4f}"
)

print(
    f"Macro Precision:"
    f" {logistic_summary['precision_macro_mean']:.4f} "
    f"± {logistic_summary['precision_macro_std']:.4f}"
)

print(
    f"Macro Recall:   "
    f"{logistic_summary['recall_macro_mean']:.4f} "
    f"± {logistic_summary['recall_macro_std']:.4f}"
)

print(
    f"Macro F1:       "
    f"{logistic_summary['f1_macro_mean']:.4f} "
    f"± {logistic_summary['f1_macro_std']:.4f}"
)

print(
    f"Weighted F1:    "
    f"{logistic_summary['f1_weighted_mean']:.4f} "
    f"± {logistic_summary['f1_weighted_std']:.4f}"
)

print("\nPer-fold Macro F1:")

for fold_number, score in enumerate(
    cv_results_logistic["test_f1_macro"],
    start=1,
):
    print(
        f"  Fold {fold_number}: "
        f"{score:.4f}"
    )

print("\n" + "=" * 70)
print("LOGISTIC REGRESSION BASELINE: COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# STEP 5.16A — XGBOOST FEATURE IMPORTANCE
# ============================================================

print("=" * 70)
print("STEP 5.16A — XGBOOST FEATURE IMPORTANCE")
print("=" * 70)

# ------------------------------------------------------------
# Fit a diagnostic copy of the final pipeline
# ------------------------------------------------------------

xgb_diagnostic_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            ColumnTransformer(
                transformers=[
                    (
                        "numerical",
                        StandardScaler(),
                        NUMERICAL_FEATURES,
                    ),
                    (
                        "categorical",
                        OneHotEncoder(
                            handle_unknown="ignore",
                            sparse_output=False,
                        ),
                        CATEGORICAL_FEATURES,
                    ),
                ],
                remainder="drop",
            ),
        ),
        (
            "model",
            XGBClassifier(
                n_estimators=300,
                max_depth=6,
                learning_rate=0.05,
                subsample=0.8,
                colsample_bytree=0.8,
                objective="multi:softprob",
                eval_metric="mlogloss",
                random_state=42,
                n_jobs=-1,
            ),
        ),
    ]
)

print("\nFitting diagnostic XGBoost on training data...")

xgb_diagnostic_pipeline.fit(
    X,
    y_encoded,
)

print("Diagnostic model fitted.")

# ------------------------------------------------------------
# Extract feature names
# ------------------------------------------------------------

diagnostic_preprocessor = (
    xgb_diagnostic_pipeline
    .named_steps["preprocessor"]
)

diagnostic_model = (
    xgb_diagnostic_pipeline
    .named_steps["model"]
)

feature_names = (
    diagnostic_preprocessor
    .get_feature_names_out()
)

importances = (
    diagnostic_model
    .feature_importances_
)

feature_importance_df = pd.DataFrame({
    "feature": feature_names,
    "importance": importances,
})

feature_importance_df = (
    feature_importance_df
    .sort_values(
        "importance",
        ascending=False,
    )
    .reset_index(drop=True)
)

print("\nTOP 20 TRANSFORMED FEATURES")
print("=" * 70)

display(
    feature_importance_df.head(20)
)

# ------------------------------------------------------------
# Aggregate importance by original feature
# ------------------------------------------------------------

def get_original_feature(transformed_name):

    if transformed_name.startswith(
        "numerical__"
    ):
        return transformed_name.replace(
            "numerical__",
            "",
        )

    if transformed_name.startswith(
        "categorical__"
    ):
        remaining = transformed_name.replace(
            "categorical__",
            "",
        )

        if remaining.startswith("Soil_"):
            return "Soil"

        if remaining.startswith("Crop_"):
            return "Crop"

    return transformed_name


feature_importance_df[
    "original_feature"
] = feature_importance_df[
    "feature"
].apply(get_original_feature)

aggregated_importance = (
    feature_importance_df
    .groupby(
        "original_feature",
        as_index=False,
    )["importance"]
    .sum()
    .sort_values(
        "importance",
        ascending=False,
    )
    .reset_index(drop=True)
)

print("\n" + "=" * 70)
print("AGGREGATED IMPORTANCE BY ORIGINAL FEATURE")
print("=" * 70)

display(aggregated_importance)

# ------------------------------------------------------------
# Save
# ------------------------------------------------------------

FEATURE_IMPORTANCE_FILE = (
    DATA_DIR
    / "fertilizer_xgboost_feature_importance.csv"
)

feature_importance_df.to_csv(
    FEATURE_IMPORTANCE_FILE,
    index=False,
)

AGGREGATED_IMPORTANCE_FILE = (
    DATA_DIR
    / "fertilizer_xgboost_feature_importance_aggregated.csv"
)

aggregated_importance.to_csv(
    AGGREGATED_IMPORTANCE_FILE,
    index=False,
)

print("\nSaved:")
print(FEATURE_IMPORTANCE_FILE)
print(AGGREGATED_IMPORTANCE_FILE)

print("\n" + "=" * 70)
print("FEATURE IMPORTANCE ANALYSIS: COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# STEP 5.16B — XGBOOST CONFIDENCE & ERROR ANALYSIS
# ============================================================

from sklearn.metrics import accuracy_score

print("=" * 70)
print("STEP 5.16B — XGBOOST CONFIDENCE & ERROR ANALYSIS")
print("=" * 70)

# ------------------------------------------------------------
# Use the already-fitted final XGBoost pipeline
# ------------------------------------------------------------

print("\nGenerating test probabilities...")

y_test_proba = xgb_pipeline.predict_proba(
    X_test_final
)

# Predicted class
y_test_pred = xgb_pipeline.predict(
    X_test_final
)

# ------------------------------------------------------------
# Confidence
# ------------------------------------------------------------

prediction_confidence = (
    np.max(
        y_test_proba,
        axis=1,
    )
)

# Second-highest probability
sorted_probabilities = np.sort(
    y_test_proba,
    axis=1,
)

second_highest_probability = (
    sorted_probabilities[:, -2]
)

confidence_margin = (
    prediction_confidence
    - second_highest_probability
)

# ------------------------------------------------------------
# Correct / incorrect
# ------------------------------------------------------------

correct_mask = (
    y_test_pred == y_test_encoded
)

incorrect_mask = ~correct_mask

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("\nPrediction confidence summary:")
print(
    f"Mean confidence:   "
    f"{prediction_confidence.mean():.4f}"
)

print(
    f"Minimum confidence:"
    f" {prediction_confidence.min():.4f}"
)

print(
    f"Maximum confidence:"
    f" {prediction_confidence.max():.4f}"
)

print(
    f"Median confidence: "
    f"{np.median(prediction_confidence):.4f}"
)

print("\nConfidence by prediction correctness:")

print(
    f"Correct predictions: "
    f"{correct_mask.sum()}"
)

print(
    f"Mean confidence for correct predictions: "
    f"{prediction_confidence[correct_mask].mean():.4f}"
)

print(
    f"Incorrect predictions: "
    f"{incorrect_mask.sum()}"
)

if incorrect_mask.sum() > 0:
    print(
        f"Mean confidence for incorrect predictions: "
        f"{prediction_confidence[incorrect_mask].mean():.4f}"
    )

# ------------------------------------------------------------
# Error table
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("INCORRECT TEST PREDICTIONS")
print("=" * 70)

error_indices = np.where(
    incorrect_mask
)[0]

error_rows = []

for idx in error_indices:

    actual_class = (
        xgb_label_encoder
        .inverse_transform(
            [y_test_encoded[idx]]
        )[0]
    )

    predicted_class = (
        xgb_label_encoder
        .inverse_transform(
            [y_test_pred[idx]]
        )[0]
    )

    error_rows.append({
        "test_row": idx,
        "actual": actual_class,
        "predicted": predicted_class,
        "confidence": prediction_confidence[idx],
        "second_best_probability":
            second_highest_probability[idx],
        "confidence_margin":
            confidence_margin[idx],
        "crop":
            X_test_final.iloc[idx]["Crop"],
        "soil":
            X_test_final.iloc[idx]["Soil"],
        "phosphorous":
            X_test_final.iloc[idx]["Phosphorous"],
        "potassium":
            X_test_final.iloc[idx]["Potassium"],
        "nitrogen":
            X_test_final.iloc[idx]["Nitrogen"],
        "carbon":
            X_test_final.iloc[idx]["Carbon"],
        "moisture":
            X_test_final.iloc[idx]["Moisture"],
        "rainfall":
            X_test_final.iloc[idx]["Rainfall"],
        "temperature":
            X_test_final.iloc[idx]["Temperature"],
        "PH":
            X_test_final.iloc[idx]["PH"],
    })

error_df = pd.DataFrame(
    error_rows
)

if len(error_df) > 0:
    display(error_df)
else:
    print("No incorrect predictions.")

# ------------------------------------------------------------
# Confidence bins
# ------------------------------------------------------------

confidence_bins = pd.cut(
    prediction_confidence,
    bins=[
        0.0,
        0.50,
        0.70,
        0.80,
        0.90,
        0.95,
        1.00,
    ],
    include_lowest=True,
)

confidence_analysis = (
    pd.DataFrame({
        "confidence_bin": confidence_bins,
        "correct": correct_mask,
    })
    .groupby(
        "confidence_bin",
        observed=False,
    )
    .agg(
        samples=("correct", "size"),
        correct=("correct", "sum"),
    )
)

confidence_analysis[
    "accuracy"
] = (
    confidence_analysis["correct"]
    / confidence_analysis["samples"]
)

print("\n" + "=" * 70)
print("CONFIDENCE BIN ANALYSIS")
print("=" * 70)

display(confidence_analysis)

# ------------------------------------------------------------
# Save error analysis
# ------------------------------------------------------------

ERROR_FILE = (
    DATA_DIR
    / "fertilizer_xgboost_test_errors.csv"
)

error_df.to_csv(
    ERROR_FILE,
    index=False,
)

CONFIDENCE_FILE = (
    DATA_DIR
    / "fertilizer_xgboost_confidence_analysis.csv"
)

confidence_analysis.to_csv(
    CONFIDENCE_FILE
)

print("\nSaved:")
print(ERROR_FILE)
print(CONFIDENCE_FILE)

print("\n" + "=" * 70)
print("CONFIDENCE & ERROR ANALYSIS: COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# STEP 5.19 — FINAL XGBOOST MODEL PACKAGING
# ============================================================

from pathlib import Path
import pandas as pd
import joblib
import json

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from xgboost import XGBClassifier


print("=" * 70)
print("STEP 5.19 — FINAL XGBOOST MODEL PACKAGING")
print("=" * 70)


# ============================================================
# 1. PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/ KrishiTrust_AI"
)

DATA_DIR = PROJECT_ROOT / "datasets"
MODEL_DIR = PROJECT_ROOT / "model"
PREPROCESSING_DIR = MODEL_DIR / "preprocessing"

MODEL_DIR.mkdir(
    parents=True,
    exist_ok=True
)

PREPROCESSING_DIR.mkdir(
    parents=True,
    exist_ok=True
)

TRAIN_FILE = DATA_DIR / "fertilizer_train.csv"
TEST_FILE = DATA_DIR / "fertilizer_test.csv"

LABEL_ENCODER_FILE = (
    DATA_DIR / "fertilizer_label_encoder.joblib"
)


# ============================================================
# 2. LOAD DATA
# ============================================================

print("\nLoading training dataset...")

train_df = pd.read_csv(TRAIN_FILE)

print(
    "Training shape:",
    train_df.shape
)

print("\nLoading testing dataset...")

test_df = pd.read_csv(TEST_FILE)

print(
    "Testing shape:",
    test_df.shape
)


# ============================================================
# 3. DEFINE FEATURES
# ============================================================

NUMERICAL_FEATURES = [
    "Temperature",
    "Moisture",
    "Rainfall",
    "PH",
    "Nitrogen",
    "Phosphorous",
    "Potassium",
    "Carbon"
]

CATEGORICAL_FEATURES = [
    "Soil",
    "Crop"
]

ALL_FEATURES = (
    NUMERICAL_FEATURES +
    CATEGORICAL_FEATURES
)

TARGET = "Fertilizer"


# ============================================================
# 4. VALIDATE COLUMNS
# ============================================================

required_columns = (
    ALL_FEATURES +
    [TARGET]
)

for column in required_columns:

    if column not in train_df.columns:
        raise ValueError(
            f"Missing training column: {column}"
        )

    if column not in test_df.columns:
        raise ValueError(
            f"Missing testing column: {column}"
        )

print("\nColumn validation: PASS")


# ============================================================
# 5. PREPARE TRAINING DATA
# ============================================================

X_train = train_df[
    ALL_FEATURES
].copy()

X_test = test_df[
    ALL_FEATURES
].copy()

y_train_text = train_df[
    TARGET
].copy()

y_test_text = test_df[
    TARGET
].copy()


# ============================================================
# 6. LOAD LABEL ENCODER
# ============================================================

print("\nLoading label encoder...")

label_encoder = joblib.load(
    LABEL_ENCODER_FILE
)

print(
    "Label encoder loaded: PASS"
)


# ============================================================
# 7. ENCODE TARGET
# ============================================================

y_train = label_encoder.transform(
    y_train_text
)

y_test = label_encoder.transform(
    y_test_text
)

print("\nTarget encoding: PASS")

print(
    "Number of classes:",
    len(label_encoder.classes_)
)

print("\nClass mapping:")

for index, class_name in enumerate(
    label_encoder.classes_
):

    print(
        f"  {index} -> {class_name}"
    )


# ============================================================
# 8. CREATE PREPROCESSOR
# ============================================================

print("\n" + "=" * 70)
print("CREATING FINAL PREPROCESSOR")
print("=" * 70)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            StandardScaler(),
            NUMERICAL_FEATURES
        ),
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            CATEGORICAL_FEATURES
        )
    ],
    remainder="drop"
)

print(
    "\nPreprocessor created."
)


# ============================================================
# 9. CREATE XGBOOST
# ============================================================

xgb_model = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="multi:softprob",
    eval_metric="mlogloss",
    random_state=42,
    n_jobs=-1
)

print("\nXGBoost configuration:")

print("  n_estimators = 300")
print("  max_depth = 6")
print("  learning_rate = 0.05")
print("  subsample = 0.8")
print("  colsample_bytree = 0.8")
print("  random_state = 42")


# ============================================================
# 10. COMPLETE PIPELINE
# ============================================================

final_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            xgb_model
        )
    ]
)

print(
    "\nComplete pipeline created."
)


# ============================================================
# 11. TRAIN FINAL MODEL
# ============================================================

print("\n" + "=" * 70)
print("TRAINING FINAL XGBOOST PIPELINE")
print("=" * 70)

print(
    "\nTraining rows:",
    len(X_train)
)

print(
    "Training features:",
    len(ALL_FEATURES)
)

print(
    "\nTraining..."
)

final_pipeline.fit(
    X_train,
    y_train
)

print(
    "\nFinal XGBoost pipeline training: COMPLETE"
)


# ============================================================
# 12. VALIDATE TRANSFORMED FEATURES
# ============================================================

transformed_features = (
    final_pipeline
    .named_steps[
        "preprocessor"
    ]
    .get_feature_names_out()
)

print(
    "\nTransformed feature count:",
    len(transformed_features)
)

if len(transformed_features) == 44:

    print(
        "Transformed feature validation: PASS"
    )

else:

    print(
        "WARNING: transformed feature count "
        f"is {len(transformed_features)}, "
        "expected 44."
    )


# ============================================================
# 13. SAVE COMPLETE PIPELINE
# ============================================================

PIPELINE_FILE = (
    MODEL_DIR /
    "fertilizer_xgboost_pipeline.joblib"
)

joblib.dump(
    final_pipeline,
    PIPELINE_FILE
)

print("\nComplete pipeline saved:")
print(
    PIPELINE_FILE
)


# ============================================================
# 14. SAVE PREPROCESSOR
# ============================================================

PREPROCESSOR_FILE = (
    PREPROCESSING_DIR /
    "fertilizer_preprocessor.joblib"
)

joblib.dump(
    preprocessor,
    PREPROCESSOR_FILE
)

print("\nPreprocessor saved:")
print(
    PREPROCESSOR_FILE
)


# ============================================================
# 15. SAVE LABEL ENCODER
# ============================================================

FINAL_LABEL_ENCODER_FILE = (
    PREPROCESSING_DIR /
    "fertilizer_label_encoder.joblib"
)

joblib.dump(
    label_encoder,
    FINAL_LABEL_ENCODER_FILE
)

print("\nLabel encoder saved:")
print(
    FINAL_LABEL_ENCODER_FILE
)


# ============================================================
# 16. SAVE FEATURE CONFIG
# ============================================================

FEATURE_CONFIG_FILE = (
    PREPROCESSING_DIR /
    "fertilizer_feature_config.json"
)

feature_config = {

    "target":
        TARGET,

    "numerical_features":
        NUMERICAL_FEATURES,

    "categorical_features":
        CATEGORICAL_FEATURES,

    "all_features":
        ALL_FEATURES,

    "excluded_features":
        ["Remark"],

    "original_feature_count":
        len(ALL_FEATURES),

    "transformed_feature_count":
        len(transformed_features),

    "number_of_classes":
        len(label_encoder.classes_),

    "classes":
        label_encoder.classes_.tolist(),

    "model":
        "XGBoost",

    "pipeline":
        True,

    "random_state":
        42
}

with open(
    FEATURE_CONFIG_FILE,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        feature_config,
        file,
        indent=4
    )

print("\nFeature configuration saved:")
print(
    FEATURE_CONFIG_FILE
)


# ============================================================
# 17. SAVE MODEL METADATA
# ============================================================

METADATA_FILE = (
    MODEL_DIR /
    "fertilizer_model_metadata.json"
)

metadata = {

    "model_name":
        "fertilizer_xgboost_pipeline",

    "purpose":
        "Fertilizer recommendation",

    "training_rows":
        int(len(X_train)),

    "testing_rows":
        int(len(X_test)),

    "number_of_classes":
        int(len(label_encoder.classes_)),

    "classes":
        label_encoder.classes_.tolist(),

    "cv_macro_f1":
        0.9921,

    "test_accuracy":
        0.9929,

    "test_macro_precision":
        0.9869,

    "test_macro_recall":
        0.9919,

    "test_macro_f1":
        0.9891,

    "test_weighted_f1":
        0.9929,

    "correct_test_predictions":
        556,

    "incorrect_test_predictions":
        4,

    "feature_count":
        len(ALL_FEATURES),

    "transformed_feature_count":
        len(transformed_features),

    "excluded_features":
        ["Remark"],

    "random_state":
        42
}

with open(
    METADATA_FILE,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )

print("\nModel metadata saved:")
print(
    METADATA_FILE
)


# ============================================================
# 18. FINAL FILE VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("FINAL PACKAGING VALIDATION")
print("=" * 70)

required_files = [

    PIPELINE_FILE,

    PREPROCESSOR_FILE,

    FINAL_LABEL_ENCODER_FILE,

    FEATURE_CONFIG_FILE,

    METADATA_FILE
]

all_files_exist = True

for file_path in required_files:

    if file_path.exists():

        size_mb = (
            file_path.stat().st_size
            / (1024 * 1024)
        )

        print(
            f"FOUND ✅  "
            f"{file_path.name} "
            f"({size_mb:.2f} MB)"
        )

    else:

        print(
            f"NOT FOUND ❌  "
            f"{file_path}"
        )

        all_files_exist = False


# ============================================================
# 19. FINAL STATUS
# ============================================================

print("\n" + "=" * 70)

if all_files_exist:

    print(
        "FINAL MODEL PACKAGING: PASS ✅"
    )

else:

    print(
        "FINAL MODEL PACKAGING: FAIL ❌"
    )

print("=" * 70)
```

```python
# ============================================================
# STEP 5.19B — NATIVE XGBOOST MODEL PACKAGING
# ============================================================

from pathlib import Path
import json
import joblib
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from xgboost import XGBClassifier


print("=" * 70)
print("STEP 5.19B — NATIVE XGBOOST MODEL PACKAGING")
print("=" * 70)


# ============================================================
# PATHS
# ============================================================

PROJECT_ROOT = Path("/content/drive/MyDrive/ KrishiTrust_AI")

DATA_DIR = PROJECT_ROOT / "datasets"
MODEL_DIR = PROJECT_ROOT / "model"
PREPROCESSING_DIR = MODEL_DIR / "preprocessing"

MODEL_DIR.mkdir(parents=True, exist_ok=True)
PREPROCESSING_DIR.mkdir(parents=True, exist_ok=True)


TRAIN_FILE = DATA_DIR / "fertilizer_train.csv"
TEST_FILE = DATA_DIR / "fertilizer_test.csv"

LABEL_ENCODER_FILE = (
    PREPROCESSING_DIR /
    "fertilizer_label_encoder.joblib"
)


NATIVE_MODEL_FILE = (
    MODEL_DIR /
    "fertilizer_xgboost.json"
)

PREPROCESSOR_FILE = (
    PREPROCESSING_DIR /
    "fertilizer_preprocessor.joblib"
)


# ============================================================
# FEATURES
# ============================================================

NUMERICAL_FEATURES = [
    "Temperature",
    "Moisture",
    "Rainfall",
    "PH",
    "Nitrogen",
    "Phosphorous",
    "Potassium",
    "Carbon",
]

CATEGORICAL_FEATURES = [
    "Soil",
    "Crop",
]

TARGET = "Fertilizer"


# ============================================================
# LOAD DATA
# ============================================================

print("\nLoading training dataset...")

train_df = pd.read_csv(TRAIN_FILE)

print("Training shape:", train_df.shape)


print("\nLoading testing dataset...")

test_df = pd.read_csv(TEST_FILE)

print("Testing shape:", test_df.shape)


# ============================================================
# PREPARE X / y
# ============================================================

X_train = train_df[
    NUMERICAL_FEATURES + CATEGORICAL_FEATURES
].copy()

X_test = test_df[
    NUMERICAL_FEATURES + CATEGORICAL_FEATURES
].copy()

y_train_text = train_df[TARGET].copy()
y_test_text = test_df[TARGET].copy()


# ============================================================
# LOAD EXISTING LABEL ENCODER
# ============================================================

print("\nLoading existing label encoder...")

label_encoder = joblib.load(LABEL_ENCODER_FILE)

y_train = label_encoder.transform(y_train_text)
y_test = label_encoder.transform(y_test_text)

print("Number of classes:", len(label_encoder.classes_))

print("\nClass mapping:")

for i, class_name in enumerate(label_encoder.classes_):
    print(f"  {i} -> {class_name}")


# ============================================================
# CREATE PREPROCESSOR
# ============================================================

print("\n" + "=" * 70)
print("CREATING PREPROCESSOR")
print("=" * 70)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            StandardScaler(),
            NUMERICAL_FEATURES,
        ),
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False,
            ),
            CATEGORICAL_FEATURES,
        ),
    ],
    remainder="drop",
)


# ============================================================
# FIT PREPROCESSOR ONLY ON TRAINING DATA
# ============================================================

print("\nFitting preprocessor on training data only...")

X_train_transformed = preprocessor.fit_transform(X_train)

X_test_transformed = preprocessor.transform(X_test)

print(
    "Transformed training shape:",
    X_train_transformed.shape
)

print(
    "Transformed testing shape:",
    X_test_transformed.shape
)


# ============================================================
# CREATE EXACT SELECTED XGBOOST MODEL
# ============================================================

print("\n" + "=" * 70)
print("CREATING FINAL XGBOOST MODEL")
print("=" * 70)

xgb_model = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1,
    objective="multi:softprob",
    num_class=len(label_encoder.classes_),
    eval_metric="mlogloss",
)


# ============================================================
# TRAIN
# ============================================================

print("\nTraining XGBoost...")

xgb_model.fit(
    X_train_transformed,
    y_train,
)

print("Training complete: PASS")


# ============================================================
# TEST PREDICTIONS
# ============================================================

print("\nGenerating test predictions...")

y_pred = xgb_model.predict(X_test_transformed)

test_accuracy = np.mean(y_pred == y_test)

print(
    f"Test accuracy from native model: "
    f"{test_accuracy:.4f}"
)


# ============================================================
# SAVE NATIVE XGBOOST MODEL
# ============================================================

print("\n" + "=" * 70)
print("SAVING NATIVE XGBOOST MODEL")
print("=" * 70)

xgb_model.save_model(
    NATIVE_MODEL_FILE
)

print(
    "Native model saved:"
)

print(NATIVE_MODEL_FILE)


# ============================================================
# SAVE PREPROCESSOR
# ============================================================

joblib.dump(
    preprocessor,
    PREPROCESSOR_FILE,
)

print(
    "\nPreprocessor saved:"
)

print(PREPROCESSOR_FILE)


# ============================================================
# CRITICAL — RELOAD NATIVE MODEL
# ============================================================

print("\n" + "=" * 70)
print("RELOAD VALIDATION")
print("=" * 70)

reloaded_model = XGBClassifier()

reloaded_model.load_model(
    NATIVE_MODEL_FILE
)

print("Native XGBoost reload: PASS ✅")


# ============================================================
# PREDICT AFTER RELOAD
# ============================================================

reloaded_predictions = reloaded_model.predict(
    X_test_transformed
)

reload_accuracy = np.mean(
    reloaded_predictions == y_test
)

print(
    f"Reloaded model accuracy: "
    f"{reload_accuracy:.4f}"
)


# ============================================================
# PREDICTION CONSISTENCY
# ============================================================

predictions_identical = np.array_equal(
    y_pred,
    reloaded_predictions
)

print(
    "\nPrediction consistency:",
    "PASS ✅" if predictions_identical else "FAIL ❌"
)


# ============================================================
# FILE VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("FINAL FILE VALIDATION")
print("=" * 70)

for file_path in [
    NATIVE_MODEL_FILE,
    PREPROCESSOR_FILE,
    LABEL_ENCODER_FILE,
]:

    if file_path.exists():

        size_mb = file_path.stat().st_size / (
            1024 * 1024
        )

        print(
            f"FOUND ✅ {file_path.name} "
            f"({size_mb:.2f} MB)"
        )

    else:

        print(
            f"NOT FOUND ❌ {file_path.name}"
        )


# ============================================================
# FINAL RESULT
# ============================================================

print("\n" + "=" * 70)
print("NATIVE XGBOOST PACKAGING RESULT")
print("=" * 70)

if (
    predictions_identical
    and abs(test_accuracy - reload_accuracy) < 1e-12
):

    print("PACKAGING VALIDATION: PASS ✅")

else:

    print("PACKAGING VALIDATION: FAIL ❌")
```
